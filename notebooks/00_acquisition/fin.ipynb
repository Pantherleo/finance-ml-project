```python
# Research review: Acquisition script requests data dynamically through yfinance. Filenames ending in present do not identify a frozen dataset version; preserve the existing input hashes before re-downloading.
import yfinance as yf
import pandas as pd

# Define the tickers
tickers = {
    "gold": "GC=F",          # Gold Futures
    "forex_usd_inr": "USDINR=X",  # USD to INR Forex Pair
    "alphabet": "GOOG",      # Alphabet (Google)
    "apple": "AAPL",         # Apple
    "crude_oil": "CL=F"      # WTI Crude Oil Futures (commodity example)
}

# Set the date range
start_date = "2007-01-01"
end_date = None  # None = fetch until today

# Download and save each dataset
for name, ticker in tickers.items():
    print(f"Downloading {name} ({ticker})...")
    data = yf.download(ticker, start=start_date, end=end_date)
    file_name = f"{name}_2007_to_present.csv"
    data.to_csv(file_name)
    print(f"Saved to {file_name}\n")

print("✅ All datasets downloaded and saved successfully!")
```
